# Inspect Agent Data

Quick visual sanity check of every CSV under `data/agent_data/` — one cell
per file, each showing `.head()`. Re-run `python fleet_agent/preprocessing.py`
first if you don't see this, or if the file list below is stale (this
notebook lists whatever `agent_data/` contains at the time it was built —
add/remove cells if the set of months changes).

```python
import sys
sys.path.append("..")

import pandas as pd
from pathlib import Path
from gps_lib import config

AGENT_DATA_DIR = Path(config.DATA_DIR) / "agent_data"
pd.set_option("display.max_columns", 25)


def head(filename: str, n: int = 5):
    path = AGENT_DATA_DIR / filename
    if not path.exists():
        print(f"{filename} not found under {AGENT_DATA_DIR} — run `python fleet_agent/preprocessing.py` first.")
        return None
    df = pd.read_csv(path, dtype={"tracker_id": str})
    print(f"{filename} — {df.shape[0]} rows x {df.shape[1]} cols")
    return df.head(n)


print("Files currently in data/agent_data/:")
for p in sorted(AGENT_DATA_DIR.glob("*.csv")):
    print(" ", p.name)
```
